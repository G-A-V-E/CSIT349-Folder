# Midterm Mini Data Science Project
## Customer Segmentation using K-Means Clustering

**Dataset:** store_customers.csv — 1,000 customers with Gender, Age, Annual Income (k$), and Spending Score (1–100)

**Research question:** Can we discover natural groups (segments) of customers based on their demographics and spending behavior, so the store can target each group with tailored marketing strategies?

**Why K-Means?** We have unlabeled data and want to find hidden group structure. K-Means partitions customers into k clusters by minimizing within-cluster variance (WCSS). It is simple, fast, and highly interpretable for segmentation — a perfect match for this dataset.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score

df = pd.read_csv('store_customers.csv')
df.head()
```

## 1. Data Preparation

The dataset contains missing values (missing Age, Income, Spending Score, and Gender entries).
Strategy:
- **Numeric columns** → impute with the **median** (robust against outliers/skew, unlike the mean).
- **Gender** → impute with the **mode** (most frequent value).
- **Standardize** the features (z-scores) so income (range 15–144) does not dominate age (range 18–80) purely because of scale — K-Means is distance-based, so scaling is essential.

```python
print("Missing values per column:")
print(df.isna().sum())

# Impute: median for numeric, mode for categorical
for col in ['Age', 'Annual Income (k$)', 'Spending Score (1-100)']:
    df[col] = df[col].fillna(df[col].median())
df['Gender'] = df['Gender'].fillna(df['Gender'].mode()[0])

print("\nAfter cleaning:", df.isna().sum().sum(), "missing values remaining")
df.describe().round(2)
```

## 2. Descriptive Statistics

```python
fig, axes = plt.subplots(1, 3, figsize=(14, 4))
features = ['Age', 'Annual Income (k$)', 'Spending Score (1-100)']
for ax, col in zip(axes, features):
    sns.histplot(df[col], kde=True, ax=ax, color='steelblue')
    ax.set_title(f'Distribution of {col}')
    ax.axvline(df[col].mean(), color='red', linestyle='--', label=f"mean={df[col].mean():.1f}")
    ax.legend()
plt.tight_layout(); plt.show()
```

## 3. Choosing k — The Elbow Method

We run K-Means for k = 1–10 and plot the **within-cluster sum of squares (inertia)**.
The "elbow" (where the curve bends sharply) indicates the best trade-off between cluster count and compactness. The elbow appears at **k = 4–5**; we also checked silhouette scores and selected **k = 5**, which yields the most interpretable, business-meaningful segments.

```python
X = df[features]
scaler = StandardScaler()
Xs = scaler.fit_transform(X)

inertias = []
for k in range(1, 11):
    km = KMeans(n_clusters=k, n_init=10, random_state=42)
    km.fit(Xs)
    inertias.append(km.inertia_)

plt.figure(figsize=(7, 4.5))
plt.plot(range(1, 11), inertias, 'bo-')
plt.xlabel('Number of clusters (k)'); plt.ylabel('Inertia (WCSS)')
plt.title('Elbow Method — Optimal k'); plt.grid(alpha=0.3); plt.show()

for k in [4, 5, 6]:
    labels = KMeans(n_clusters=k, n_init=10, random_state=42).fit_predict(Xs)
    print(f"k={k}: silhouette score = {silhouette_score(Xs, labels):.3f}")
```

## 4. Running K-Means (k = 5)

```python
kmeans = KMeans(n_clusters=5, n_init=10, random_state=42)
df['Cluster'] = kmeans.fit_predict(Xs)

names = {0: 'Affluent Mature (low spend)', 1: 'Young Low-Income Spenders',
         2: 'Average Everything', 3: 'Wealthy Seniors (minimal spend)',
         4: 'Young Mid-Income Engaged'}
df['Segment'] = df['Cluster'].map(names)

profile = df.groupby('Cluster')[features].mean().round(1)
profile['Size'] = df['Cluster'].value_counts().sort_index()
profile['% Female'] = (df.groupby('Cluster')['Gender']
                       .apply(lambda s: (s == 'F').mean()) * 100).round(1)
profile
```

## 5. Visualizing the Clusters

```python
fig, ax = plt.subplots(figsize=(9, 6))
palette = sns.color_palette('tab10', 5)
for c in range(5):
    sub = df[df['Cluster'] == c]
    ax.scatter(sub['Annual Income (k$)'], sub['Spending Score (1-100)'],
               s=40, alpha=0.7, color=palette[c], label=names[c])
centers = scaler.inverse_transform(kmeans.cluster_centers_)
ax.scatter(centers[:, 1], centers[:, 2], s=300, marker='X', c='black',
           label='Centroids', zorder=5)
ax.set_xlabel('Annual Income (k$)'); ax.set_ylabel('Spending Score (1-100)')
ax.set_title('K-Means Customer Segments (k=5)')
ax.legend(fontsize=8); ax.grid(alpha=0.3); plt.show()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(14, 4))
for ax, col in zip(axes, features):
    prof = df.groupby('Segment')[col].mean().reindex(list(names.values()))
    prof.plot(kind='barh', ax=ax, color='teal', alpha=0.8)
    ax.set_title(f'Mean {col} by Segment'); ax.invert_yaxis()
    ax.tick_params(axis='y', labelsize=7)
plt.tight_layout(); plt.show()
```

## 6. Insights & Business Recommendations

| Segment | Profile | Size | Recommendation |
|---|---|---|---|
| **Young Low-Income Spenders** | Age ~26, income $26k, **spending 62** | 218 | Highest engagement relative to income — student discounts, installment plans, budget lines |
| **Young Mid-Income Engaged** | Age ~33, income $42k, **spending 53** | 315 | Largest segment — loyalty programs, cross-sell, upsell |
| **Average Everything** | Age ~39, income $63k, spending 42 | 191 | Mid engagement — targeted promotions to lift spend |
| **Affluent Mature (low spend)** | Age ~47, income **$86k**, spending 25 | 160 | **Biggest opportunity**: high income, low spend — premium products, personalized service, win-back campaigns |
| **Wealthy Seniors (minimal spend)** | Age ~68, income **$109k**, spending **4** | 116 | Nearly disengaged — low priority; simple essentials or deprioritize marketing spend |

**Key findings:**
1. **Age matters more than gender** — the gender split is ~50/50 in every cluster, so gender is not a useful segmentation variable here.
2. There is a striking **negative relationship between income and spending score** among older customers: the two richest segments (avg $86k and $109k) are the two least engaged.
3. The most valuable growth opportunity is converting the **Affluent Mature** segment — they have money but currently don't spend it at this store.
4. The youngest segment already spends enthusiastically despite the lowest income — strong brand affinity among young adults.
